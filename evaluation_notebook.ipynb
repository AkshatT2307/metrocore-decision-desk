## Scoring

Your answer **A** is compared to the reference optimal value **A\***.

| Your answer | Score |
|:---|---:|
| Correct: \|A − A\*\| / max(1, \|A\*\|) ≤ 0.01 | **+1** |
| Close: within 5% of A\* | **+0.5** |
| Wrong number (feasible problem, off by > 5%) | **0** |
| Wrong number on infeasible (should be `na`) | **−0.5** |
| `na` on a feasible problem | **−0.5** |
| Correct `na` on infeasible | **+1** |
| Missing answer or crash | **0** |

**Final score** = mean of per-problem scores across the hidden evaluation set.
Blank answers and ordinary per-problem exceptions score 0. Invalid answers also score 0; only `na` means infeasible. Structural submission errors are rejected. A fatal notebook failure may invalidate the Kaggle submission.
